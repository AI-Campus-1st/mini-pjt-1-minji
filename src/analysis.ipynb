```python
# maria DB 에 store_csv 파일 연결 

import pandas as pd
import pymysql
import os
import numpy as np
from dotenv import load_dotenv

load_dotenv()


store_df = pd.read_csv('../data/processed/base_df.csv', encoding='utf-8')
conn = pymysql.connect(
        host = os.environ.get('DB_HOST'),
        port = int (os.environ.get('DB_PORT', 3306)),
        user=os.environ.get('DB_USER'),
        password=os.environ.get ('DB_PASSWORD'),
        database=os.environ.get ('DB_NAME'),
        charset='utf8mb4' )
sql = """
INSERT INTO tb_store(
    store_id,
    store_name,
    branch_name,
    category_large,
    category_medium,
    category_small,
    province_name,
    city_name,
    district_name)
VALUES (%s, %s, %s, %s, %s, %s, %s, %s, %s)
"""

store_df = store_df.replace({np.nan: None}) #NaN을 None으로 변환

data = store_df[[
    '상가업소번호',
    '상호명',
    '지점명',
    '상권업종대분류명',
    '상권업종중분류명',
    '상권업종소분류명',
    '시도명',
    '시군구명',
    '행정동명'
]].values.tolist()



with conn.cursor() as cur:
    cur.executemany(sql, data)

conn.commit()
# conn.close()
```

```python
# maria DB 에 store_csv 파일 연결 

import pandas as pd
import pymysql
import os
import numpy as np
from dotenv import load_dotenv

load_dotenv()


population_df = pd.read_csv('../data/processed/he_conn_df.csv', encoding='utf-8')
conn = pymysql.connect(
        host = os.environ.get('DB_HOST'),
        port = int (os.environ.get('DB_PORT', 3306)),
        user=os.environ.get('DB_USER'),
        password=os.environ.get ('DB_PASSWORD'),
        database=os.environ.get ('DB_NAME'),
        charset='utf8mb4' )
sql = """
INSERT INTO tb_population(
    province_name,
    city_name,
    district_name,
    age,
    male_population,
    female_population)
VALUES (%s, %s, %s, %s, %s, %s)
"""

population_df = population_df.replace({np.nan: None}) #NaN을 None으로 변환

data = population_df[[
    '시도명',
    '시군구명',
    '읍면동명',
    '나이',
    '남자 인구수',
    '여자 인구수'
]].values.tolist()



with conn.cursor() as cur:
    cur.executemany(sql, data)

conn.commit()
# conn.close()
```

```python
# 파생 지표 계산
store_df = pd.read_csv('../data/processed/base_df.csv', encoding='utf-8')
population_df = pd.read_csv('../data/processed/he_conn_df.csv', encoding='utf-8')

store_group = store_df.groupby('시군구명').size()
#store_group
population_df['총 인구수'] = population_df['남자 인구수'] + population_df['여자 인구수']
population_group = population_df.groupby('시군구명')['총 인구수'].sum()
df_concat_sp = pd.concat([store_group, population_group], axis=1)
df_concat_sp.columns = ['지역별 상가수', '지역별 인구수']

```

```python
# 지역별 인구수를 기준으로 내림차순
df_concat_sp.sort_values(by='지역별 인구수', ascending=False)
```

```python
# 인구가 만명으로 동일할 때 상가수가 어디가 제일 많은지 확인
df_concat_sp['인구 1만명당 상가수'] = (df_concat_sp['지역별 상가수'] / df_concat_sp['지역별 인구수'] * 10000)
df_concat_sp.sort_values(by='인구 1만명당 상가수', ascending=False)
```

```python
# 1. 평균 구하기
mean_store = df_concat_sp['인구 1만명당 상가수'].mean().round(2)
mean_store
```

```python
# Numpy 벡터 연산
import numpy as np
df_concat_sp['평균'] = np.where (
    df_concat_sp['인구 1만명당 상가수'] >= mean_store,
    '평균 이상',
    '평균 이하 '
)

df_concat_sp
```
